# Triangular → Regular grid interpolation example

```python
import numpy as np
import matplotlib.pyplot as plt
from mesh_interpolation_tools.tri2reg import tri2reg
from scipy.spatial import Delaunay
```

# 1. Generate random triangular mesh

```python
np.random.seed(0)
points = np.random.rand(200, 2)
tri = Delaunay(points)
nodes = points.T  # shape (2, N)
elem = tri.simplices.T  # shape (3, M)
```


# 2. Define a synthetic scalar field on nodes

```python
field_at_nodes = np.sin(np.pi * nodes[0]) * np.cos(np.pi * nodes[1])
```

# 3. Define regular grid

```python
grid_x = np.linspace(0, 1, 100)
grid_y = np.linspace(0, 1, 100)
```

# 4. Perform interpolation

```python
gx, gy, heatmap = tri2reg(nodes, elem, field_at_nodes, grid_x, grid_y)
```

# 5. Plot results

```python
plt.figure(figsize=(12,5))

plt.subplot(121)
plt.tricontourf(nodes[0], nodes[1], elem.T, field_at_nodes, cmap="viridis")
plt.title("Field on triangular mesh")
plt.colorbar()

plt.subplot(122)
plt.pcolormesh(gx, gy, heatmap.T, shading="auto", cmap="viridis")
plt.title("Interpolated on regular grid")
plt.colorbar()

plt.tight_layout()
plt.show()
```
