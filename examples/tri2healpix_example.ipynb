# Spherical triangular mesh → HEALPix interpolation

```python
import numpy as np
import healpy as hp
import matplotlib.pyplot as plt
from mesh_interpolation_tools.tri2healpix import conservative_interp_spherical_tri2healpix
from scipy.spatial import SphericalVoronoi
```

# 1. Create random points on sphere (unit sphere)

```python
np.random.seed(3)
phi = 2*np.pi*np.random.rand(500)
theta = np.arccos(2*np.random.rand(500)-1)
x = np.sin(theta)*np.cos(phi)
y = np.sin(theta)*np.sin(phi)
z = np.cos(theta)
nodes_src = np.vstack([x, y, z])
```

# Build a simple triangulation using convex hull

```python
from scipy.spatial import ConvexHull
hull = ConvexHull(nodes_src.T)
elem_src = hull.simplices.T
```

# 2. Define field (e.g., f = z-coordinate)

```python
field_src = nodes_src[2]
```

# 3. HEALPix target grid

```python
nside = 32
```

# 4. Interpolate

```python
healpix_map = conservative_interp_spherical_tri2healpix(nodes_src, elem_src, field_src, nside)
```

# 5. Plot

```python
hp.mollview(healpix_map, title="Interpolated field on HEALPix grid", cmap="coolwarm")
hp.graticule()
plt.show()
```
