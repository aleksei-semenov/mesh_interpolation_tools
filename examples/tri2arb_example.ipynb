# Triangular → Non-uniform structured grid interpolation

```python
import numpy as np

import matplotlib.pyplot as plt
from mesh_interpolation_tools.tri2arb import tri2arb
from scipy.spatial import Delaunay
```

# 1. Generate triangular mesh

```python
np.random.seed(1)
points = np.random.rand(200, 2)
tri = Delaunay(points)
nodes = points.T
elem = tri.simplices.T
```

# 2. Field

```python
field_at_nodes = np.exp(-((nodes[0]-0.5)**2 + (nodes[1]-0.5)**2) * 20)
```

# 3. Define non-uniform grid

```python
grid_x = np.sort(np.random.rand(30))
grid_y = np.sort(np.random.rand(40))
```

# 4. Interpolate

```python
gx, gy, heatmap = tri2arb(nodes, elem, field_at_nodes, grid_x, grid_y)
```

# 5. Plot

```python
plt.figure(figsize=(12,5))

plt.subplot(121)
plt.tricontourf(nodes[0], nodes[1], elem.T, field_at_nodes, cmap="plasma")
plt.title("Field on triangular mesh")
plt.colorbar()

plt.subplot(122)
plt.pcolormesh(gx, gy, heatmap.T, shading="auto", cmap="plasma")
plt.title("Interpolated on irregular grid")
plt.colorbar()

plt.tight_layout()
plt.show()
```
