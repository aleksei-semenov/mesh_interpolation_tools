# Triangular → Triangular interpolation

```python
import numpy as np
import matplotlib.pyplot as plt
from mesh_interpolation_tools.tri2tri import conservative_interp_tri2tri
from scipy.spatial import Delaunay
```

# 1. Source triangular mesh

```python
np.random.seed(2)
points_src = np.random.rand(100, 2)
tri_src = Delaunay(points_src)
nodes_src = points_src.T
elem_src = tri_src.simplices.T
field_src = np.sin(2*np.pi*nodes_src[0]) * np.sin(2*np.pi*nodes_src[1])
```

# 2. Target triangular mesh

```python
points_tgt = np.random.rand(200, 2)
tri_tgt = Delaunay(points_tgt)
nodes_tgt = points_tgt.T
elem_tgt = tri_tgt.simplices.T
```

# 3. Interpolate

```python
field_tgt = conservative_interp_tri2tri(nodes_src, elem_src, field_src, nodes_tgt, elem_tgt)
```

# 4. Plot

```python
plt.figure(figsize=(12,5))

plt.subplot(121)
plt.tricontourf(nodes_src[0], nodes_src[1], elem_src.T, field_src, cmap="viridis")
plt.title("Source mesh field")
plt.colorbar()

plt.subplot(122)
plt.tricontourf(nodes_tgt[0], nodes_tgt[1], elem_tgt.T, field_tgt, cmap="viridis")
plt.title("Interpolated on target mesh")
plt.colorbar()

plt.tight_layout()
plt.show()
```
